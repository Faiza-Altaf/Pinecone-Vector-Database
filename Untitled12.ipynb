```python
# CELL 2: Setup Pinecone API Key & Model
from pinecone import Pinecone, ServerlessSpec
from sentence_transformers import SentenceTransformer

# Double quotes ke andar apni copied key paste karein
PINECONE_API_KEY = "pcsk_5BDGG_Gi7Bf8KyNhHwedAbJLAkmKkzapeJU4VZ9m6xn1dnpKGFW5b41e8ck9AUAwZY6Lc"

pc = Pinecone(api_key=PINECONE_API_KEY)
model = SentenceTransformer('all-MiniLM-L6-v2')
embedding_dim = model.get_embedding_dimension()
print(f"Model loaded! Dimension: {embedding_dim}")
```

```python
# ==========================================
# CELL 3: Create Pinecone Index
# ==========================================
index_name = "assignment-10-vector-db"

# Check if index exists, create if not
existing_indexes = [index.name for index in pc.list_indexes()]

if index_name not in existing_indexes:
    pc.create_index(
        name=index_name,
        dimension=embedding_dim,
        metric="cosine",
        spec=ServerlessSpec(
            cloud="aws",
            region="us-east-1"
        )
    )
    print(f"Index '{index_name}' successfully create ho gaya hai!")
else:
    print(f"Index '{index_name}' pehle se mojood hai.")

# Connect to index
index = pc.Index(index_name)
```

```python
documents = [
    {"id": "vec1", "text": "Artificial intelligence is transforming software engineering and product design.", "category": "tech"},
    {"id": "vec2", "text": "Vector databases store dense numeric representations of text, images, and audio.", "category": "database"},
    {"id": "vec3", "text": "Pinecone allows fast similarity search over high-dimensional vector embeddings.", "category": "database"},
    {"id": "vec4", "text": "Frontend development with React and Tailwind CSS enables rapid UI prototyping.", "category": "web_dev"},
    {"id": "vec5", "text": "Machine learning models require clean preprocessing and feature extraction.", "category": "tech"}
]

vectors_to_upsert = []
for doc in documents:
    vector_values = model.encode(doc["text"]).tolist()
    vectors_to_upsert.append({
        "id": doc["id"],
        "values": vector_values,
        "metadata": {"text": doc["text"], "category": doc["category"]}
    })

upsert_response = index.upsert(vectors=vectors_to_upsert)
print("Upsert Completed Successfully!")
print("Index Statistics:", index.describe_index_stats())
```

```python
query_text = "How do vector databases handle similarity search?"
query_embedding = model.encode(query_text).tolist()

results = index.query(
    vector=query_embedding,
    top_k=3,
    include_metadata=True
)

print(f"Search Query: '{query_text}'\n")
print("Top Relevant Matches:")
for match in results["matches"]:
    print(f"• [Score: {match['score']:.4f}] {match['metadata']['text']} (Category: {match['metadata']['category']})")
```
